```python
import numpy as np
import matplotlib.pyplot as plt
import super_resolution as sr
from super_resolution import cartoon_upsampling_4x
from super_resolution import cartoon_upsampling_8x
from PIL import Image
```

```python
# resize image (or just crop)
img = Image.open('input.png')
```

```python
plt.imshow(np.asarray(img))
```

```python
img2 = np.array(img)
```

```python
img_small = img2[500:1012,500:1012]
```

```python
plt.imshow(img_small)
```

```python
# convert smaller crop back to image
im = Image.fromarray(img_small)
im.save("crop512.jpeg")
```

```python
plt.imshow(img_small[0:125,375:500])
```

## 4x upsampling

```python
image_4x = cartoon_upsampling_4x('./input.png', './images/input_4x.png' )
```

```python
plt.imshow(image_4x)
```

```python
plt.imshow(image_4x[0:500,1500:2000])
```

```python
image_8x = cartoon_upsampling_8x( './crop512.jpeg', './images/crop512_8x.jpg' )
```

```python
plt.imshow(image_8x[0:1000,3000:4000])
```

```python
plt.figure(figsize=(16,8))
plt.subplot(1,2,1)
plt.imshow(img_small)
plt.subplot(1,2,2)
plt.imshow(image_8x)
```

```python
# more zoom to see details
r0,r1 = 200,300 # orig image coords
c0,c1 = 375,500

plt.figure(figsize=(16,8))
plt.subplot(1,2,1)
plt.imshow(img_small[r0:r1,c0:c1])
plt.subplot(1,2,2)
plt.imshow(image_8x[r0*8:r1*8,c0*8:c1*8])
```

```python
# more zoom to see details
r0,r1 = 1,125 # orig image coords
c0,c1 = 375,500

plt.figure(figsize=(16,8))
plt.subplot(1,2,1)
plt.imshow(img_small[r0:r1,c0:c1])
plt.title('Original')
plt.subplot(1,2,2)
plt.imshow(image_8x[r0*8:r1*8,c0*8:c1*8])
plt.title('8x upsample with Neural Net');
```
